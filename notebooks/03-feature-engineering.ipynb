## Feature Engineering Summary

The cleaned dataset was enhanced with two meaningful features:

* **`mileage_per_year`** — captures vehicle usage relative to its age.
* **`brand_model`** — combines the standardized brand and model into a specific vehicle identifier.

The final feature dataset contains **97,675 records and 14 columns**.

Target-derived fields such as `price_log` and `price_segment` will not be used as input features for price prediction to avoid target leakage.
